# Pancreas: verified cross-study reference

This notebook fetches the official Zenodo archive to a caller-selected cache, materializes the source-provided raw-count combined matrix, and explicitly holds out one study. It does not use the archive's aligned matrix.

```python
import os
from pathlib import Path

import bioml_data as bio

data_dir = Path(os.environ.get("BIOML_DATA_DIR", ".cache/bioml-data"))
held_out_study = os.environ.get("BIOML_HELD_OUT_STUDY", "Muraro")
raw = bio.download_dataset("pancreas-four-study", data_dir=data_dir).artifact
prepared = bio.prepare_dataset("pancreas-four-study", artifact=raw, data_dir=data_dir)
dataset = bio.load_dataset("pancreas-four-study", artifact=prepared.lineage)
fold = bio.pancreas_lodo_split(dataset, held_out_study=held_out_study)
report = bio.pancreas_metadata_concordance(dataset, held_out_study=held_out_study)
print(dataset.counts.shape)
print(held_out_study, sum(item.partition == "test" for item in fold.assignments))
[
    (
        item.partition,
        item.observation_count,
        [comparison.status for comparison in item.comparisons],
    )
    for item in report.partition_reports
]
```
